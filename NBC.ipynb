# ***- Preprocessing***

**CLEANNING**

```python
URL_RE = re.compile(r"(https?://\S+|www\.\S+)", flags=re.IGNORECASE)

def cleaning_stage(text: str) -> str:
    """Cleaning: hapus URL, @mention, #hashtag, rapikan spasi"""
    if pd.isna(text):
        return ""
    raw = str(text)

    raw = URL_RE.sub(" ", raw)             # hapus URL
    raw = re.sub(r"@\S+", " ", raw)        # hapus mention
    raw = re.sub(r"#\S+", " ", raw)        # hapus hashtag
    raw = re.sub(r"\s+", " ", raw).strip() # rapikan spasi

    return raw
```

```python
stemmer = StemmerFactory().create_stemmer()

STOPWORDS_ID = set(stopwords.words("indonesian"))
NEGATION_KEEP = {"tidak", "bukan", "jangan", "tak"}
STOPWORDS_ID = STOPWORDS_ID - NEGATION_KEEP

URL_RE = re.compile(r"(https?://\S+|www\.\S+)", flags=re.IGNORECASE)

# 1) Normalisasi slang/variasi ejaan
NORM_DICT = {
    "tp": "tapi",
    "tdk": "tidak",
    "gk": "tidak",
    "ga": "tidak",
    "gak": "tidak",
    "nggak": "tidak",
    "s7": "setuju",

    "goks": "gokil",
    "yesss": "yes",
    "yess": "yes",
    "yeess": "yes",
    "gg": "bagus",
    "dmn": "",

    "nyekekkkk": "mencekik",
    "nyekekkk": "mencekik",
    "nyekek": "mencekik",

    "gacorrr": "gacor",
    "gacorr": "gacor",

    "colapse": "kolaps",
    "collapse": "kolaps",
    "kolaps": "kolaps",

    "koboy": "pemberani",
    "coboy": "pemberani",
    "coby": "pemberani",

    "bpk": "Bapak",
    "lucknat": "laknat",
}

# 2) Noise tokens (hapus)
NOISE_WORDS = {
    "wk", "wkwk", "wkwkwk", "haha", "hahaha", "lol",
    "tf", "sini",
    "kacau",
}

# 3) Frasa/ungkapan netral (hint netral)
NEUTRAL_PHRASES = [
    "tak kenal maka tak sayang",
    "rakyat jelata",
    "ga paham",
    "gak paham",
    "hati hati",
    "hati-hati",
]

# 4) Emoji handling:
EMOJI_TO_TOKEN = {
    # positif dasar
    "👍": "emojisetuju",
    "👏": "emojiplause",
    "😍": "emojisuka",
    "🔥": "emojisemangat",
    "❤️": "emojicinta",

    "✨": "emojibagus",
    "🌟": "emojibagus",

    "👌": "emojisetujupos",
    "🙌": "emojihebatpos",
    "🤝": "emojisepakatpos",

    # negatif
    "👎": "emojitidaksetuju",
    "🤢": "emojijijik",
    "🥴": "emojisakit",
}

POS_EMOJI_TOKENS = {
    "emojisetuju", "emojiplause", "emojisuka", "emojisemangat", "emojicinta",
    "emojibagus", "emojisetujupos", "emojihebatpos", "emojisepakatpos",
}
NEG_EMOJI_TOKENS = {"emojitidaksetuju", "emojijijik", "emojisakit"}

# Regex emoji (untuk hapus emoji lain yang tersisa)
EMOJI_RE = re.compile(
    "["
    "\U0001F300-\U0001F5FF"
    "\U0001F600-\U0001F64F"
    "\U0001F680-\U0001F6FF"
    "\U0001F700-\U0001F77F"
    "\U0001F780-\U0001F7FF"
    "\U0001F800-\U0001F8FF"
    "\U0001F900-\U0001F9FF"
    "\U0001FA00-\U0001FA6F"
    "\U0001FA70-\U0001FAFF"
    "\u2600-\u26FF"
    "\u2700-\u27BF"
    "]+",
    flags=re.UNICODE
)

def replace_emojis_keep_known(text: str) -> str:

    if not isinstance(text, str):
        text = "" if text is None else str(text)

    # 1) ubah emoji yang dikenali menjadi token
    for e, tok in EMOJI_TO_TOKEN.items():
        text = text.replace(e, f" {tok} ")

    # 2) hapus emoji lain yang tersisa
    text = EMOJI_RE.sub(" ", text)

    # rapikan spasi
    text = re.sub(r"\s+", " ", text).strip()
    return text

# 5) Lexicon hint berbasis istilah (untuk teks pendek)
NEG_HINT_TERMS = {
    "dpr",
    "omon",
    "riba",
    "fiktif",
    "baju orange",
    "mencekik",
    "susah",
    "halu",
    "kolaps",
    "boneka",
    "tong kosong nyaring bunyinya",
    "seenak jidat",
    "sesukanya",
    "lawak",
}

POS_HINT_TERMS = {
    "setuju",
    "semoga",
    "senang",
    "pinter",
    "cuan",
    "keren",
    "yes",
    "betul",
    "jenius",
    "sangat menarik",
    "gacor",
    "gokil",
    "bagus",
    "let him cook",
    "memasak",
    "meme masak",
    "pemberani",
}

# 6) Deteksi pertanyaan
QUESTION_WORDS = {"apa", "kenapa", "mengapa", "gimana", "bagaimana", "kapan", "siapa", "dimana", "berapa"}

# 7) Hapus bahasa asing
EN_STOP = {
    "the","and","or","but","to","of","in","on","for","with","is","are","was","were",
    "be","been","it","this","that","you","we","they","i","me","my","your","our",
    "not","no","yes","ok","okay","good","game","cook","him","let"
}

def remove_english_tokens(tokens):
    return [t for t in tokens if t not in EN_STOP]

# 8) Drop komentar yang hanya emoji/hint
EMOJI_TOKENS = set(EMOJI_TO_TOKEN.values())
HINT_TOKENS = {"hintnegatif", "hintpositif", "hintnetral", "hintnetralphrase"}

def is_only_emoji_or_hint(tokens):
    return all((t in EMOJI_TOKENS) or (t in HINT_TOKENS) for t in tokens) and len(tokens) > 0
```

**CASE FOLDING**

```python
def case_folding_stage(text: str) -> str:
    if pd.isna(text):
        return ""
    return str(text).lower()
```

**NORMALISASI**

```python
def normalize_laughter(text: str) -> str:
    text = re.sub(r"\b(wk){2,}\b", " ", text)
    text = re.sub(r"\b(wkwk){1,}\b", " ", text)
    text = re.sub(r"\b(ha){2,}\b", " ", text)
    return text

def squash_repeats(text: str) -> str:
    return re.sub(r"(.)\1{3,}", r"\1\1", text)

def normalisasi_stage(text: str) -> str:
    if not text:
        return ""
    raw = str(text)

    # 1) normalisasi tawa dan repetisi karakter
    raw = normalize_laughter(raw)
    raw = squash_repeats(raw)

    # 2) emoji: yang dikenali -> token, lainnya -> dihapus
    raw = replace_emojis_keep_known(raw)

    # 3) angka -> token
    raw = re.sub(r"\d+", " angka ", raw)

    # 4) bersihkan selain huruf/_/spasi
    raw = re.sub(r"[^a-z_\s]", " ", raw)

    # 5) rapikan spasi
    raw = re.sub(r"\s+", " ", raw).strip()
    return raw

```

**TOKENIZING**

```python
import numpy as np
def tokenizing_stage(x):
    i = 0
    x['token']=np.zeros[len(x)]
    for txt in x['clean']:
      txt.split(' ')
      x['token'][i] = txt
      i += 1
    return x
```

**STOP WORD REMOVAL**

```python
def stopword_removal_stage(tokens, raw_lower: str):
    if not tokens:
        return []

    # deteksi pertanyaan
    is_question = any(q in raw_lower for q in QUESTION_WORDS)

    toks = []
    for t in tokens:
        t = NORM_DICT.get(t, t)
        if t in NOISE_WORDS:
            continue
        if t in STOPWORDS_ID:
            continue
        toks.append(t)

    toks = remove_english_tokens(toks)
    toks = [t for t in toks if len(t) >= 2]

    # sentiment hints
    has_strong = False

    for p in NEUTRAL_PHRASES:
        if p in raw_lower:
            toks.append("hintnetralphrase")

    if any(t in POS_EMOJI_TOKENS for t in toks):
        toks.append("hintpositif"); has_strong = True
    if any(t in NEG_EMOJI_TOKENS for t in toks):
        toks.append("hintnegatif"); has_strong = True

    for term in NEG_HINT_TERMS:
        if term in raw_lower:
            toks.append("hintnegatif"); has_strong = True
            break
    for term in POS_HINT_TERMS:
        if term in raw_lower:
            toks.append("hintpositif"); has_strong = True
            break

    if is_question and not has_strong:
        toks.append("hintnetral")

    return toks

def STP_remove(text: str) -> str:
    if not isinstance(text, str) or text.strip() == "":
        return ""

    # NORMALISASI
    raw = text.lower()

    # simpan raw_lower untuk rule-based sentiment
    raw_lower = raw

    # TOKENISASI
    tokens = raw.split()

    # STOPWORD REMOVAL
    tokens_final = stopword_removal_stage(tokens, raw_lower)

    # OUTPUT
    return " ".join(tokens_final)
```

**STEMMING**

```python
def stemming_stage(text: str) -> str:
    if not isinstance(text, str) or text.strip() == "":
        return ""
    out = stemmer.stem(text)
    out = re.sub(r"\s+", " ", out).strip()
    return out
```

```python
nltk.download('stopwords')

stopwords_id = set(stopwords.words('indonesian'))

stopwords_wc = set(STOPWORDS) | stopwords_id
stopwords_wc.update([
    'https', 'http', 'www', 'com', 'co', 'id',
    'rt', 'amp', 'yg', 'ga', 'gk', 'nggak',
    'tdk', 'tp', 'nih', 'sih', 'deh',
    'lah', 'loh', 'kok', 'lu', 'gue', 'gw',
    'aja', 'udah', 'sudah', 'dah', 'nya',
    'dll', 'kalo', 'kalau', 'jadi', 'buat',
    'bikin', 'biar', 'angka'
])

def wc_clean_text(text: str) -> str:
    text = str(text).lower()
    text = re.sub(r"(https?://\S+|www\.\S+)", " ", text)
    text = re.sub(r"@\w+", " ", text)
    text = re.sub(r"#\w+", " ", text)
    text = re.sub(r"[^a-zA-Z\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

text_before = " ".join(
    df["Comment"]
    .dropna()
    .astype(str)
    .map(wc_clean_text)
    .tolist()
)

wc_before = WordCloud(
    stopwords=stopwords_wc,
    background_color="white",
    max_words=500,
    width=800,
    height=400
).generate(text_before)

text_after = " ".join(
    df["clean"]
    .dropna()
    .astype(str)
    .map(wc_clean_text)
    .tolist()
)

wc_after = WordCloud(
    stopwords=stopwords_wc,
    background_color="white",
    max_words=500,
    width=800,
    height=400
).generate(text_after)

plt.figure(figsize=(16, 8))

plt.subplot(1, 2, 1)
plt.imshow(wc_before, interpolation="bilinear")
plt.axis("off")
plt.title("WordCloud Before Preprocessing", fontsize=16)

plt.subplot(1, 2, 2)
plt.imshow(wc_after, interpolation="bilinear")
plt.axis("off")
plt.title("WordCloud After Preprocessing", fontsize=16)

plt.tight_layout()
plt.show()
```
